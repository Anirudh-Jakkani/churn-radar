# Telco Churn: Exploratory Data Analysis
A quick look at the data before modeling. The interactive version lives in the dashboard's **Data Explorer** tab.

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from src.data import load_raw, clean
from src.features import add_features
sns.set_theme(style='whitegrid')
df = add_features(clean(load_raw()))
df.shape
```

```python
df.isna().sum().sum(), df['Churn'].value_counts(normalize=True)
```

## Churn rate by key segments

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 8))
for ax, col in zip(axes.flat, ['Contract', 'tenure_bucket', 'InternetService', 'PaymentMethod']):
    df.groupby(col)['Churn'].mean().sort_values().plot.barh(ax=ax, color='#8B5CF6')
    ax.set_title(f'Churn rate by {col}'); ax.set_xlabel('')
plt.tight_layout()
```

## Numeric features: churned vs stayed

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, col in zip(axes, ['tenure', 'MonthlyCharges', 'TotalCharges']):
    sns.kdeplot(data=df, x=col, hue='Churn', common_norm=False, fill=True, ax=ax)
plt.tight_layout()
```

```python
sns.heatmap(df[['tenure','MonthlyCharges','TotalCharges','avg_monthly_spend','num_services','Churn']].corr(), annot=True, fmt='.2f', cmap='vlag', center=0)
```

**Takeaways:** month-to-month contracts, short tenure, fiber internet, and electronic-check payment are strongly linked to churn. Long tenure and 2-year contracts protect against it. Classes are imbalanced (~26.5% churn), so we evaluate with recall / F1 / ROC-AUC rather than accuracy.
